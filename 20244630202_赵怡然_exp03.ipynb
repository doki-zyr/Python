### 实验1：正则表达式的基础匹配

```python
import re

# 任务1：基础字符匹配
print("========== 任务1：基础字符匹配 ==========")

text1 = "cat on mat; later: cats and dogs"
patterns1 = {
    "数字 (\\d+)": r"\d+",
    "空白字符 (\\s+)": r"\s+",
    "独立单词 cat (\\bcat\\b)": r"\bcat\b"
}

for name, pattern in patterns1.items():
    result = re.findall(pattern, text1)
    print(f"{name}：{result}")
```

```python
# ========== 任务2：选择符匹配 ==========
print("\n========== 任务2：选择符匹配 ==========")
text2 = "cat dog bird CAT"
print("cat|dog ->", re.findall(r"cat|dog", text2))
print("a|aa|aaa ->", re.findall(r"a|aa|aaa", "aaaa"))
print("aaa|aa|a ->", re.findall(r"aaa|aa|a", "aaaa"))
print("说明：同一匹配起点按 | 左侧到右侧依次尝试，先成功的分支先被采用。")
```

```python
print("\n========== 任务3：量词匹配 ==========")
text3 = "ct cat caat caaat"
patterns3 = {
    r"ca*t": "*表示出现 0 次或多次",
    r"ca+t": "+表示出现 1 次或多次",
    r"ca?t": "?表示出现 0 次或 1 次",
    r"ca{2}t": "{2}表示恰好出现 2 次",
    r"ca{1,2}t": "{1,2}表示出现 1～2 次",
}
for pattern, meaning in patterns3.items():
    print(f"{pattern:9s} -> {re.findall(pattern, text3)}；{meaning}")
```

### 实验2：分组、回溯引⽤与替换

```python
# ========== 任务1：邮箱分组提取 ==========
print("========== 任务1：邮箱分组提取 ==========")
email_text = "联系：alice@example.com；备选：bob@test.org"
email_pattern = r"([A-Za-z0-9._%+-]+)@([A-Za-z0-9.-]+\.[A-Za-z]{2,})"
email_matches = re.findall(email_pattern, email_text)
for username, domain in email_matches:
    print(f"用户名：{username}；域名：{domain}")
```

```python
print("\n========== 任务2：后向引用 ==========")
repeat_text = "cat cat, dog cat, bird bird"
repeat_pattern = r"\b(\w+)\s+\1\b"
repeat_matches = list(re.finditer(repeat_pattern, repeat_text))
for match in repeat_matches:
    print(f"重复文本：{match.group(0)}；重复单词：{match.group(1)}")
```

```python
print("\n========== 任务3：日期格式转换 ==========")
date_text = "2024-05-01\n2024/05/02\n2024.05.03"
date_pattern = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
converted_dates = re.sub(date_pattern, r"\2-\3-\1", date_text)
print("转换前：\n" + date_text)
print("转换后：\n" + converted_dates)
print("转换前行数：", len(date_text.splitlines()))
print("转换后行数：", len(converted_dates.splitlines()))
assert len(date_text.splitlines()) == len(converted_dates.splitlines())
```

```python
print("\n========== 任务4：电话号码格式化 ==========")
phone_text = "联系我：123-456-7890 或 987.654.3210"
phone_pattern = r"\b(\d{3})[-.](\d{3})[-.](\d{4})\b"
converted_phones = re.sub(phone_pattern, r"(\1) \2-\3", phone_text)
print("转换前：", phone_text)
print("转换后：", converted_phones)
```

### 实验3：使⽤ BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# ========== 任务1：构建解析树和读取标题 ==========
print("========== 任务1：构建解析树 ==========")
soup = BeautifulSoup(html_doc, "html.parser")
print("页面标题文本：", soup.title.get_text(strip=True))
h1 = soup.find("h1")
print("h1 标签名：", h1.name)
print("h1 标签文本：", h1.get_text(strip=True))
```

```python
# ========== 任务2：段落与链接 ==========
print("\n========== 任务2：查找指定标签 ==========")
course_paragraphs = soup.find_all("p", class_="course")
print(f"course 段落数量：{len(course_paragraphs)}")
for i, paragraph in enumerate(course_paragraphs, start=1):
    print(f"第{i}个课程段落：{paragraph.get_text(strip=True)}")

links = soup.find_all("a")
print(f"链接数量：{len(links)}")
for i, link in enumerate(links, start=1):
    print(f"第{i}个链接：文字={link.get_text(strip=True)}，href={link.get('href')}")
```

```python
# ========== 任务3：比较 get_text() 的范围 ==========
print("\n========== 任务3：比较 get_text() 结果 ==========")
print("整个文档文本：", soup.get_text(separator=" | ", strip=True))
print("页面 body 文本：", soup.body.get_text(separator=" | ", strip=True))
print("单个段落文本：", course_paragraphs[0].get_text(strip=True))
print("单个链接文本：", links[0].get_text(strip=True))
print("说明：文档文本包含 title 和正文；body 仅包含正文；段落、链接分别仅提取标签内部文本。")
```
