## 一、Python 正则表达式练习

### 1. 检查字符串是否仅包含英文字母 a-z、A-Z 和数字 0-9。

```python
import re
```

```python
def check_string(s):
    return bool(re.fullmatch(r'[a-zA-Z0-9]+', s))


# 测试用例
print(check_string("abcABC123"))
print(check_string("Python2026"))
print(check_string("abc@123"))
print(check_string("Hello World"))
```

### 2. 匹配一个 a 后面跟着零个或多个 b 的字符串。

```python
def match_ab_zero(s):
    return bool(re.fullmatch(r'ab*', s))


# 测试用例
print(match_ab_zero("a"))
print(match_ab_zero("ab"))
print(match_ab_zero("abbb"))
print(match_ab_zero("ac"))
```

### 3. 匹配一个 a 后面跟着一个或多个 b 的字符串。

```python
def match_ab_one(s):
    return bool(re.fullmatch(r'ab+', s))


# 测试用例
print(match_ab_one("a"))
print(match_ab_one("ab"))
print(match_ab_one("abbb"))
print(match_ab_one("ac"))
```

### 4. 匹配以 a 开头、以 b 结尾，且中间可以有任意字符的字符串。

```python
def match_start_end(s):
    return bool(re.fullmatch(r'a.*b', s, flags=re.DOTALL))


# 测试用例
print(match_start_end("ab"))
print(match_start_end("a123b"))
print(match_start_end("a hello b"))
print(match_start_end("abc"))
print(match_start_end("cab"))
```

### 5. 删除 IP 地址中各段数字的前导零。

例如，192.168.001.010 应变为 192.168.1.10；某段全为零时保留一个 0。

```python
def remove_ip_zeros(s):
    return re.sub(r'\b0+(?=\d)', '', s)


# 测试用例
print(remove_ip_zeros("192.168.001.010"))
print(remove_ip_zeros("010.000.002.003"))
print(remove_ip_zeros("000.000.000.000"))
print(remove_ip_zeros("192.168.1.1"))
```

### 6. 将字符串中的空格替换为下划线，并将下划线替换为空格。

```python
def swap_space_underscore(s):
    return re.sub(r'[ _]', lambda m: '_' if m.group() == ' ' else ' ', s)


# 测试用例
print(swap_space_underscore("Hello World"))
print(swap_space_underscore("Hello_World"))
print(swap_space_underscore("Python is_easy"))
```

### 7. 将 yyyy-mm-dd 格式的日期转换为 dd-mm-yyyy 格式。

```python
def convert_date(s):
    return re.sub(r'(\d{4})-(\d{2})-(\d{2})', r'\3-\2-\1', s)


# 测试用例
print(convert_date("2026-11-24"))
print(convert_date("2025-01-08"))
print(convert_date("2006-12-31"))
```

### 8. 找出给定字符串中所有以 a 或 e 开头的单词。

```python
def find_ae_words(s):
    return re.findall(r'\b[ae][a-zA-Z]*\b', s)


# 测试用例
print(find_ae_words("apple banana elephant orange egg"))
print(find_ae_words("an animal eats an apple"))
print(find_ae_words("Python Java Linux"))
```

### 9. 将给定字符串中的 Road 缩写为 Rd。

```python
def replace_road(s):
    return re.sub(r'\bRoad\b', 'Rd', s)


# 测试用例
print(replace_road("123 Main Road"))
print(replace_road("Park Road and Green Road"))
print(replace_road("Roadside"))
```

### 10. 找出字符串中所有长度至少为 4 个字符的单词。

```python
def find_long_words(s):
    return re.findall(r'\b[a-zA-Z]{4,}\b', s)


# 测试用例
print(find_long_words("I love learning Python"))
print(find_long_words("cat dog book apple"))
print(find_long_words("a an the"))
```

### 11. 删除字符串中除字母和数字之外的所有字符。

```python
def remove_special_chars(s):
    return re.sub(r'[^a-zA-Z0-9]', '', s)


# 测试用例
print(remove_special_chars("Hello, World! 123"))
print(remove_special_chars("Python_2026@#"))
print(remove_special_chars("!@#$%"))
```

### 12. 删除字符串中的括号及括号内的内容。

对下面的输入逐项处理：

- example (.com)
- w3resource
- github (.com)
- stackoverflow (.com)

```python
def remove_parentheses(s):
    return re.sub(r'\([^()]*\)', '', s).strip()


# 测试用例
strings = [
    "example (.com)",
    "w3resource",
    "github (.com)",
    "stackoverflow (.com)"
]

for s in strings:
    print(remove_parentheses(s))
```

## 二、从⽹⻚检索⽓候数据

```python
import requests
from bs4 import BeautifulSoup
from datetime import date
```

### 1.⽓候数据表格的城市⻚⾯

```python
url = "https://en.wikipedia.org/wiki/Madison,_Wisconsin"

headers = {
    "User-Agent": "Mozilla/5.0"
}

# 获取网页并检查请求是否成功
response = requests.get(url, headers=headers, timeout=30)
response.raise_for_status()

# 解析 HTML
soup = BeautifulSoup(response.text, "html.parser")

print("访问 URL：", url)
print("访问日期：", date.today())

# 查找气候数据表格
climate_table = None

for table in soup.find_all("table"):
    for tag in table.find_all(["th", "caption"]):
        title = " ".join(tag.get_text().split())

        if title.startswith("Climate data for"):
            climate_table = table
            break

    if climate_table is not None:
        break

# 观察表格节点结构
if climate_table is not None:
    children = climate_table.find_all(recursive=False)

    print("表格标题：", title)
    print("表格节点标签：", climate_table.name)
    print("直接子节点数量：", len(children))
    print("直接子节点标签名：", [child.name for child in children])
    print("表格总行数：", len(climate_table.find_all("tr")))
else:
    print("未找到气候数据表格")
```

**表格结构说明：**

通过 BeautifulSoup 成功定位城市气候数据表格，其节点标签为
`table`。该表格包含一个直接子标签 `tbody`，用于存放表格主体。
`tbody` 中包含多个 `tr` 标签，每个 `tr` 表示表格的一行，
其中 `th` 用于表示标题或行名，`td` 用于存放具体的气候数据。

### 2.编写 retrieve_climate_table(s) & 多个城市⻚⾯测试

```python
def retrieve_climate_table(s):
    headers = {"User-Agent": "Mozilla/5.0"}

    response = requests.get(s, headers=headers, timeout=30)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    for table in soup.find_all("table"):

        # 情况1：标题位于 caption 或 th 标签中
        for tag in table.find_all(["caption", "th"]):
            title = " ".join(tag.get_text().split())

            if title.startswith("Climate data for"):
                return table

        # 情况2：标题位于表格内部的其他标签中
        table_text = " ".join(table.get_text(" ", strip=True).split())

        if table_text.startswith("Climate data for"):
            return table

        # 情况3：标题位于表格前面的标题标签中
        previous = table.find_previous_sibling()

        if previous is not None:
            title = " ".join(previous.get_text().split())

            if title.startswith("Climate data for"):
                return table

    return None
```

```python
urls = {
    "Madison": "https://en.wikipedia.org/wiki/Madison,_Wisconsin",
    "Buenos Aires": "https://en.wikipedia.org/wiki/Buenos_Aires",
    "Harbin": "https://en.wikipedia.org/wiki/Harbin",
    "Los Angeles": "https://en.wikipedia.org/wiki/Los_Angeles",
    "Boston": "https://en.wikipedia.org/wiki/Boston"
}

for city, url in urls.items():
    print("城市：", city)
    print("URL：", url)
    print("访问日期：", date.today())

    try:
        table = retrieve_climate_table(url)

        if table is not None:
            print("返回类型：", type(table))
            print("节点标签：", table.name)

            children = table.find_all(recursive=False)
            print("直接子节点数量：", len(children))
            print("直接子节点标签：", [tag.name for tag in children])

            caption = table.find("caption")
            first_row = table.find("tr")

            if caption is not None:
                print("Caption：", caption.get_text(strip=True))

            if first_row is not None:
                print("第一行：", " ".join(first_row.get_text().split()))

        else:
            print("未找到气候数据表格")

    except requests.RequestException as e:
        print("网页访问失败：", e)

    print("-" * 60)
```

```python
url = "https://en.wikipedia.org/wiki/Python_(programming_language)"

print("访问 URL：", url)
print("访问日期：", date.today())

try:
    print(retrieve_climate_table(url))
except requests.RequestException as e:
    print("网页访问失败：", e)
```

### 3. 编写 list_climate_table_row_names(s)

```python
def list_climate_table_row_names(s):
    table = retrieve_climate_table(s)

    if table is None:
        return None

    row_names = []

    for row in table.find_all("tr"):

        # 获取当前行的直接单元格
        cells = row.find_all(["th", "td"], recursive=False)

        # 没有数据单元格的行不处理
        if not any(cell.name == "td" for cell in cells):
            continue

        # 优先查找 scope="row" 的行标题
        row_header = row.find("th", attrs={"scope": "row"},
                              recursive=False)

        # 如果没有 scope="row"，则使用第一个 th
        if row_header is None:
            row_header = next(
                (cell for cell in cells if cell.name == "th"),
                None
            )

        if row_header is None:
            continue

        # 提取纯文本并清理空白
        name = " ".join(row_header.get_text().split()).strip()

        # 排除标题行和数据来源行
        if not name:
            continue

        if name.lower().startswith(
            ("climate data for", "month", "source")
        ):
            continue

        row_names.append(name)

    return row_names
```

```python
test_cities = {
    "Madison": "https://en.wikipedia.org/wiki/Madison,_Wisconsin",
    "Buenos Aires": "https://en.wikipedia.org/wiki/Buenos_Aires",
    "Harbin": "https://en.wikipedia.org/wiki/Harbin",
    "Los Angeles": "https://en.wikipedia.org/wiki/Los_Angeles",
    "Boston": "https://en.wikipedia.org/wiki/Boston"
}

for city, url in test_cities.items():
    print("城市：", city)

    try:
        table = retrieve_climate_table(url)
        names = list_climate_table_row_names(url)

        print("是否找到气候表格：", table is not None)
        print("返回类型：", type(table) if table is not None else None)
        print("表格总行数：",
              len(table.find_all("tr")) if table is not None else None)
        print("提取的行名：", names)
        print("行名数量：", len(names) if names is not None else None)
        print("是否全部为字符串：",
              all(isinstance(x, str) for x in names)
              if names is not None else None)
        print("是否去除首尾空白：",
              all(x == x.strip() for x in names)
              if names is not None else None)

    except requests.RequestException as e:
        print("网页访问失败：", e)

    print("-" * 60)
```

```python
url = "https://en.wikipedia.org/wiki/Python_(programming_language)"

print("访问 URL：", url)
print("访问日期：", date.today())

try:
    print(list_climate_table_row_names(url))
except requests.RequestException as e:
    print("网页访问失败：", e)
```
