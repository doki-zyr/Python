```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError(f"分数必须在 0~100 之间，收到 {score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

for sc in (95, 85, 72, 59, 60, 89, 90):
    print(f"{sc:>3} -> {score_level(sc)}")
```

```python
def factorial_recursive(n):
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:                  # 基线条件，约定 0! = 1
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

for n in (0, 1, 5, 10):
    r, it = factorial_recursive(n), factorial_iterative(n)
    print(f"{n:>2}! -> 递归 {r:>10} 迭代 {it:>10} 一致 {r == it}")
```

```python
def summary(values):
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)

low, high, avg = summary([85, 58, 92, 76, 60, 89])   
print(f"min={low} max={high} mean={avg:.4f}")
```

```python
import os

print("当前工作目录:", os.getcwd())
DATA_DIR = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(DATA_DIR, exist_ok=True)
TXT_PATH = os.path.join(DATA_DIR, "students.txt")
```

```python
RAW = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(TXT_PATH, "w", encoding="utf-8") as f:
    f.write(RAW)

with open(TXT_PATH, encoding="utf-8") as f:
    header = f.readline().strip()      
    for line in f:                    
        print(repr(line.strip()))
```

```python
def parse_student(line):
    """把一行 'name,score' 解析成字典；无效记录抛 ValueError。"""
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(fields)}")

    name, score_text = fields[0].strip(), fields[1].strip()
    if not name:
        raise ValueError("姓名为空")

    try:
        score = int(score_text)
    except ValueError as exc:
        raise ValueError(f"成绩不是整数: {score_text!r}") from exc

    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围: {score}")

    return {"name": name, "score": score}
```

```python
valid_records, invalid_records = [], []

with open(TXT_PATH, encoding="utf-8") as f:
    f.readline()                                  # 跳过表头
    for lineno, line in enumerate(f, start=2):    # 数据从第 2 行开始
        if not line.strip():                      # 跳过空行
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:
            invalid_records.append((lineno, line.strip(), str(exc)))

print(f"有效 {len(valid_records)} 条")
for lineno, raw, reason in invalid_records:
    print(f"  第 {lineno} 行 {raw!r} 无效 -> {reason}")
```

```python
n_valid, n_invalid = len(valid_records), len(invalid_records)
avg_score = sum(r["score"] for r in valid_records) / n_valid
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / n_valid

REPORT_PATH = os.path.join(DATA_DIR, "report.txt")

with open(REPORT_PATH, "w", encoding="utf-8") as f:      # w：清空重写
    f.write(f"有效记录数: {n_valid}\n")
    f.write(f"无效记录数: {n_invalid}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.1%}\n")

print(open(REPORT_PATH, encoding="utf-8").read())

with open(REPORT_PATH, "a", encoding="utf-8") as f:
    f.write("备注：a 模式追加的一行\n")
with open(REPORT_PATH, encoding="utf-8") as f:
    print(f.read())
```

| 模式 | 打开时对已有内容的处理 | 写指针位置 | 演示结果 |
|---|---|---|---|
| `w` | **立即把文件截断为 0 字节**，原有内容全部丢失 | 文件开头 | 原有两行消失，只剩新写入的内容 |
| `a` | **原样保留** | 文件末尾 | 原有两行保留，新内容追加在最后 |

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name!r}, score={self.score}, {status})"
    
students = [Student(r["name"], r["score"]) for r in valid_records]
for st in students:
    print(st, "| is_passed():", st.is_passed())
```

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)        # 复用父类初始化
        self.research_topic = research_topic

    def __str__(self):                       # 重写
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"topic={self.research_topic!r})")

g = GraduateStudent("Alice", 85, "机器学习")
print(g, "| is_passed():", g.is_passed())    # 父类方法直接可用
```

```python
import copy

def fresh():
    return {"name": "Alice", "scores": [85, 90]}

# 直接赋值
direct = fresh()
a = direct
a["scores"].append(100)
print("直接赋值 -> 原对象:", direct)
# 浅拷贝
shallow = fresh()
b = copy.copy(shallow)
b["scores"].append(100)
print("copy.copy -> 原对象:", shallow)
# 深拷贝
deep = fresh()
c = copy.deepcopy(deep)
c["scores"].append(100)
print("deepcopy -> 原对象:", deep)
```
