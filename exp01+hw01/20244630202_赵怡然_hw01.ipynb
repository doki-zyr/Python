```python
import os, uuid, hashlib, platform, getpass

# ===== 1. 填写学号 =====
STUDENT_ID = 20244630202         

# ===== 2. 打印 conda 环境 =====
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV", "base")
print(f"[Conda 环境] {CONDA_ENV}")

# ===== 3. 机器指纹（绑定当前电脑）=====
MACHINE_FP = hashlib.sha256(
    (
        f"{platform.node()}|{platform.machine()}|{platform.system()}|"
        f"{uuid.getnode()}"
    ).encode()
).hexdigest()[:16]

# ===== 4. 生成唯一 ID =====
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"
print(f"[学号]     {STUDENT_ID}")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID]   {RUN_ID}")
```

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m
for a, b in [(2018, 2019), (1200, 300), (5040, 60)]:
    print(f"gcd({a}, {b}) = {gcd(a, b)}")
```

```python
for a, b in [(-2018, 2019), (2018, -2019), (-2018, -2019), (-48, 18), (48, -18)]:
    print(f"gcd({a}, {b}) = {gcd(a, b)}")

print("Python 的取模符号规则：-2018 % 2019 =", -2018 % 2019, "，48 % -18 =", 48 % -18)
```

没有意义。最大公约数的数学定义是"能同时整除两个数的**最大的正整数**"。返回 `-6` 时，虽然 -6 也能整除 48 和 -18，但 6 比 -6 大，所以 -6 不满足"最大"这个条件。这个负号纯粹是取模运算符号规则带来的副作用，不是算法本身的语义。

```python
def euler_limit(x):    
    return (1 + 1 / x) ** x
```

```python
for x in (1, 2, 10, 100, 1000, 100000):
    print(f"euler_limit({x:>6}) = {euler_limit(x)!r}")
```

```python
def euler_infinite_sum(n):
    total = 0.0
    term = 1.0                 
    for k in range(n):
        total += term
        term /= (k + 1)         
    return total

for n in (0, 1, 2, 3, 4, 5, 10):
    print(f"euler_infinite_sum({n:>2}) = {euler_infinite_sum(n)!r}")
```

```python
import math

def euler_approx(eps):
    total = 0.0
    term = 1.0
    k = 0
    while abs(total - math.exp(1)) > eps:
        total += term         
        k += 1
        term /= k              
    return total
```

```python
def euler_approx_debug(eps):
    total, term, k = 0.0, 1.0, 0
    while abs(total - math.exp(1)) > eps:
        total += term
        k += 1
        term /= k
    return total, k

for eps in (1, 0.1, 0.01, 1e-6, 1e-12):
    val, used = euler_approx_debug(eps)
    print(f"eps={eps:g}  返回={val!r}  用了 {used} 项  误差={abs(val - math.exp(1)):.3e}")
```

```python
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print_euler_sum_table(6)
```

```python
def is_power_of_2(n):
    while n > 1 and n % 2 == 0:
        n = n // 2
    return n == 1
for n in (1, 2, 3, 4, 5, 6, 8, 16, 1024, 2047):
    print(f"is_power_of_2({n:>5}) = {is_power_of_2(n)}")
```

```python
def is_power(b, n):
    while n > 1 and n % b == 0:
        n = n // b
    return n == 1
```

```python
for b, n in [(2, 8), (2, 6), (3, 27), (3, 26), (10, 1), (10, 0), (10, 100), (5, 125), (4, 16)]:
    print(f"is_power({b:>2}, {n:>4}) = {is_power(b, n)}")
```
