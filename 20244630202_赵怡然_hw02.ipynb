### 字符串的乐趣

```python
def is_palindrome(text):
    normalized = text.replace(" ", "").lower()
    return normalized == normalized[::-1]

print("--- 任务1：回文判断测试 ---")
print(f"is_palindrome('tacocat') 结果: {is_palindrome('tacocat')}")
print(f"is_palindrome('T A C O cat') 结果: {is_palindrome('T A C O cat')}")
print(f"is_palindrome('rats live on no evil star') 结果: {is_palindrome('rats live on no evil star')}")
print(f"is_palindrome('hello') 结果: {is_palindrome('hello')}")
```

```python
def is_abecedarian(word):
    """判断忽略空格后的字母是否按字母顺序排列。"""
    letters = word.replace(" ", "").lower() 
    return all(letters[i] <= letters[i + 1] for i in range(len(letters) - 1))

print("\n--- 任务2：字母顺序判断测试 ---")
print(f"is_abecedarian('adder') 结果: {is_abecedarian('adder')}")
print(f"is_abecedarian('beet') 结果: {is_abecedarian('beet')}")
print(f"is_abecedarian('abcd efgh xyz') 结果: {is_abecedarian('abcd efgh xyz')}")
print(f"is_abecedarian('dog') 结果: {is_abecedarian('dog')}")
print(f"is_abecedarian('cat') 结果: {is_abecedarian('cat')}")
```

```python
def remove_vowels(text):
    """返回删掉所有元音字母后的字符串。"""
    return "".join(char for char in text if char.lower() not in "aeiou")

print("\n--- 任务3：移除元音测试 ---")
print(f"remove_vowels('cat') 结果: '{remove_vowels('cat')}'")
print(f"remove_vowels('audio') 结果: '{remove_vowels('audio')}'")
print(f"remove_vowels('aeiou') 结果: '{remove_vowels('aeiou')}'")
print(f"remove_vowels('AEIOU') 结果: '{remove_vowels('AEIOU')}'")
```

### 列表的乐趣

```python
def list_reverse(items):
    """返回列表的反转副本。"""
    if not isinstance(items, list):
        raise TypeError("items must be a list")
    return items[::-1]

print("--- 任务1：列表反转测试 ---")
print(f"list_reverse([1, 2, 3]) 结果: {list_reverse([1, 2, 3])}")
print(f"list_reverse([]) 结果: {list_reverse([])}")
print(f"list_reverse([1]) 结果: {list_reverse([1])}")

# 测试异常抛出
print("\n测试非列表输入 (1, 2, 3):")
try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print(f"成功捕获异常: TypeError: {e}")
```

```python
def is_sorted(seq):
    """判断序列是否按升序排列，只检查一次相邻元素。"""
    for index in range(len(seq) - 1):
        if seq[index] > seq[index + 1]:
            return False
    return True

print("\n--- 任务2：升序判断测试 ---")
print(f"is_sorted([1, 2, 2, 3]) 结果: {is_sorted([1, 2, 2, 3])}")
print(f"is_sorted([]) 结果: {is_sorted([])}")
print(f"is_sorted([1]) 结果: {is_sorted([1])}")
print(f"is_sorted('abc') 结果: {is_sorted('abc')}")
print(f"is_sorted([1, 3, 2]) 结果: {is_sorted([1, 3, 2])}")
```

### 作为向量的元组

```python
def vec_scalar_mult(t, s):
    # 检查输入类型
    if not isinstance(t, tuple):
        raise TypeError("参数 t 必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("参数 s 必须是整数或浮点数")
    # 检查元组内元素是否都是数字
    for val in t:
        if not isinstance(val, (int, float)):
            raise TypeError("元组 t 中的元素必须是数字")
            
    # 返回新元组
    return tuple(val * s for val in t)

print("--- 任务1：向量与标量相乘测试 ---")
print(f"vec_scalar_mult((1, 2, 3), 2) 结果: {vec_scalar_mult((1, 2, 3), 2)}")
print(f"vec_scalar_mult((1.5, 2.0), 3) 结果: {vec_scalar_mult((1.5, 2.0), 3)}")

print("\n测试异常情况：")
try:
    vec_scalar_mult([1, 2, 3], 2) # 故意传入列表
except TypeError as e:
    print(f"成功捕获异常: {e}")
```

```python
def vec_inner_product(x, y):
    # 检查输入类型
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("参数 x 和 y 必须是元组")
    # 检查长度是否一致
    if len(x) != len(y):
        raise ValueError("两个元组的长度必须一致")
        
    # 计算内积 (利用 zip 将对应元素打包，再用 sum 求和)
    # 题目要求返回浮点数，所以用 float() 强转一下
    return float(sum(a * b for a, b in zip(x, y)))

print("\n--- 任务2：向量内积测试 ---")
print(f"vec_inner_product((1, 2, 3), (4, 5, 6)) 结果: {vec_inner_product((1, 2, 3), (4, 5, 6))}") # 1*4+2*5+3*6 = 32
print(f"vec_inner_product((1.5, 2), (2, 1.5)) 结果: {vec_inner_product((1.5, 2), (2, 1.5))}")

print("\n测试异常情况：")
try:
    vec_inner_product((1, 2), (1, 2, 3)) # 故意长度不一致
except ValueError as e:
    print(f"成功捕获异常: {e}")
```

```python
def check_valid_mx(mx):
    # 最外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 空元组通常不被视为有效矩阵（因为没有列）
    if len(mx) == 0:
        return False
        
    # 1. 外层元组的每个元素本身都是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
            
    # 2. 所有内层元组长度相同
    first_row_len = len(mx[0])
    for row in mx:
        if len(row) != first_row_len:
            return False
            
    # 3. 每个内层元组的每个元素都是整数或浮点数
    for row in mx:
        for val in row:
            if not isinstance(val, (int, float)):
                return False
                
    return True

print("\n--- 任务3：有效矩阵检查测试 ---")
print(f"check_valid_mx(((1, 2), (3, 4))) 结果: {check_valid_mx(((1, 2), (3, 4)))}") # True
print(f"check_valid_mx(((1, 2), (3, 4, 5))) 结果: {check_valid_mx(((1, 2), (3, 4, 5)))}") # False (长度不同)
print(f"check_valid_mx(((1, 2), [3, 4])) 结果: {check_valid_mx(((1, 2), [3, 4]))}") # False (内层有列表)
print(f"check_valid_mx(((1, 'a'), (3, 4))) 结果: {check_valid_mx(((1, 'a'), (3, 4)))}") # False (有字符)
```

```python
def mx_vec_mult(mx, v):
    # 使用任务3的函数检查矩阵
    if not check_valid_mx(mx):
        raise ValueError("无效的矩阵格式")
    if not isinstance(v, tuple):
        raise TypeError("向量 v 必须是元组")
        
    # 检查维度是否匹配：矩阵的列数(第一行的长度)必须等于向量的长度
    if len(mx[0]) != len(v):
        raise ValueError("矩阵的列数与向量的长度不匹配，无法相乘")
        
    # 计算：遍历矩阵的每一行，与向量 v 做内积
    result = []
    for row in mx:
        dot_product = sum(a * b for a, b in zip(row, v))
        result.append(dot_product)
        
    return tuple(result)

print("\n--- 任务4：矩阵与向量相乘测试 ---")
matrix = ((1, 2), (3, 4), (5, 6))
vector = (2, 3)

print(f"矩阵 {matrix} 乘 向量 {vector} 结果: {mx_vec_mult(matrix, vector)}")

print("\n测试异常情况：")
try:
    mx_vec_mult(((1, 2), (3, 4)), (1, 2, 3)) # 列数不匹配
except ValueError as e:
    print(f"成功捕获异常: {e}")
```

### 更多向量的乐趣

```python
class Vector:
    def __init__(self, dimension, elements=None):
        # 1. 验证维度是否为正整数或非负整数（这里包含0，题目说非负整数）
        if not isinstance(dimension, int) or dimension < 0:
            raise ValueError("维度必须是非负整数")
        
        self.dimension = dimension
        
        # 2. 处理默认值
        if elements is None:
            self.elements = tuple([0] * dimension)
        else:
            # 3. 验证元素列表/元组长度与维度是否一致
            if len(elements) != dimension:
                raise ValueError(f"维度 {dimension} 与元素个数 {len(elements)} 不一致")
            
            # 4. 验证元素类型是否为整数或浮点数
            for elem in elements:
                if not isinstance(elem, (int, float)):
                    raise TypeError("元素必须是整数或浮点数")
            
            self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dimension}, {list(self.elements)})"

# --- 任务 1 测试用例 ---
print("--- 任务1：测试正常情况 ---")
v1 = Vector(2, [4, 5])
v2 = Vector(3) # 测试默认值
print(f"v1: {v1}")
print(f"v2 (默认值): {v2}")

print("\n--- 任务1：测试异常情况 ---")
try:
    Vector(-1, [1, 2])
except ValueError as e:
    print(f"捕获异常 (负数维度): {e}")

try:
    Vector(2, [1, 2, 3])
except ValueError as e:
    print(f"捕获异常 (维度不一致): {e}")

try:
    Vector(2, [1, "a"])
except TypeError as e:
    print(f"捕获异常 (类型错误): {e}")
```

**任务2说明：**
维度和元素应该是**实例属性**（Instance Attributes），而不是类属性（Class Attributes）。

**理由：**
1. 每个 `Vector` 对象都有自己独立的维度大小和具体的坐标元素。如果设为类属性，所有实例将共享同一个维度和元素列表，这违背了向量对象的实际物理意义（例如二维向量和三维向量不能共享同一个维度值）。
2. 实例属性通过 `self.dimension` 和 `self.elements` 定义，能够保证每个对象的数据相互隔离，互不影响。

```python
class Vector(Vector):
    # 辅助方法：检查类型和维度
    def _check_compare(self, other):
        if not isinstance(other, Vector):
            raise TypeError("比较对象必须也是 Vector 类型")
        if self.dimension != other.dimension:
            raise ValueError("不能比较维度不同的两个向量")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dimension != other.dimension:
            raise ValueError("不能比较维度不同的两个向量")
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_compare(other)
        # 直接利用元组的字典序比较特性
        return self.elements < other.elements

    def __le__(self, other):
        self._check_compare(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_compare(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_compare(other)
        return self.elements >= other.elements

# --- 任务 3 测试用例 ---
print("\n--- 任务3：测试比较运算符 ---")
v_a = Vector(2, [2, 4])
v_b = Vector(2, [2, 5])
v_c = Vector(2, [2, 4])

print(f"v_a == v_c: {v_a == v_c}") # True
print(f"v_a < v_b: {v_a < v_b}")   # True (按字典序比较)
print(f"v_a <= v_c: {v_a <= v_c}") # True
print(f"v_b > v_a: {v_b > v_a}")   # True
print(f"v_b >= v_a: {v_b >= v_a}") # True

print("\n测试比较异常情况:")
try:
    v_a < Vector(3, [1, 2, 3])
except ValueError as e:
    print(f"捕获异常 (维度不同): {e}")
```

```python
class Vector(Vector):
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 类型")
        if self.dimension != other.dimension:
            raise ValueError("向量维度不一致，无法计算内积")
        
        # 计算对应元素乘积的总和
        result = sum(a * b for a, b in zip(self.elements, other.elements))
        return result

# --- 任务 4 测试用例 ---
print("\n--- 任务4：测试向量内积 ---")
v1 = Vector(2, [2, 4])
v2 = Vector(2, [3, 5])
print(f"v1.dot(v2): {v1.dot(v2)}") 

print("\n测试内积异常情况:")
try:
    v1.dot(Vector(3, [1, 2, 3]))
except ValueError as e:
    print(f"捕获异常 (维度不一致): {e}")
try:
    v1.dot(100)  
except TypeError as e:
    print(f"捕获异常 (类型错误): {e}")
```

```python
class Vector(Vector):
    def __mul__(self, other):
        # 1. 标量乘向量
        if isinstance(other, (int, float)):
            new_elements = [x * other for x in self.elements]
            return Vector(self.dimension, new_elements)
        
        # 2. 向量乘向量（逐元素相乘）
        elif isinstance(other, Vector):
            if self.dimension != other.dimension:
                raise ValueError("向量维度不一致，无法进行逐元素相乘")
            new_elements = [a * b for a, b in zip(self.elements, other.elements)]
            return Vector(self.dimension, new_elements)
        
        else:
            raise TypeError("乘法对象必须是数字或 Vector 类型")

    def __rmul__(self, other):

        return self.__mul__(other)

# --- 任务 5 测试用例 ---
print("\n--- 任务5：测试乘法运算符 ---")
v = Vector(3, [2, 2, 3])
v_other = Vector(3, [4, 5, 1])

# 测试标量乘法
print(f"2 * v: {2 * v}")    # 标量左乘
print(f"v * 2: {v * 2}")    # 标量右乘

# 测试向量逐元素乘法
print(f"v * v_other: {v * v_other}") # 期望结果：(8, 10, 3)

print("\n测试乘法异常情况:")
try:
    v * Vector(2, [1, 2])
except ValueError as e:
    print(f"捕获异常 (维度不一致): {e}")
```
